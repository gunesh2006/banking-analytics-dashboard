```python
#recency
def sort(group):
    return group.sort_values(['year','month_no','day_no'],ascending=[False,False,False])
latest_transactions=success_tr.groupby(['customer_id','customer_name']).apply(sort).groupby(level=["customer_id", "customer_name"]).head(1)
latest_transactions['recency']=(pd.Timestamp.today()-latest_transactions['transaction_date']).dt.days
latest_transactions
```

```python
#frequency
frequency=success_tr.groupby(['customer_id','customer_name'])['transaction_id'].count()
frequency
```

```python
#monetary
monetary=success_tr.groupby(['customer_id','customer_name'])['amount'].sum()
monetary


```

```python
rfm=latest_transactions.merge(
    frequency,
    on=['customer_id','customer_name']
).merge(
    monetary,
    on=['customer_id','customer_name']
)

rfm.rename(
    columns={'transaction_id_y':'frequency','amount_y':'monetary'},
    inplace=True
)
rfm
```

```python
rfm['R_score']=pd.qcut(
    rfm['recency'],
    5,
    labels=[5,4,3,2,1]
)
rfm['F_score'] = pd.qcut(
    rfm['frequency'],
    5,
    labels=[1, 2, 3, 4, 5]
)
rfm['M_score'] = pd.qcut(
    rfm['monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)
rfm['RFM_Score']=rfm['R_score'].astype(str)+rfm['F_score'].astype(str)+rfm['M_score'].astype(str)
rfm
```

```python
rfm['segment'] = np.select(
    [
        (rfm['R_score'].astype(int) >= 4) &
        (rfm['F_score'].astype(int) >= 4) &
        (rfm['M_score'].astype(int) >= 4),

        (rfm['R_score'].astype(int) >= 3) &
        (rfm['F_score'].astype(int) >= 4) &
        (rfm['M_score'].astype(int) >= 3),

        (rfm['R_score'].astype(int) >= 4) &
        (rfm['F_score'].astype(int) <= 3),

        (rfm['R_score'].astype(int) <= 2) &
        (rfm['F_score'].astype(int) >= 3),

        (rfm['R_score'].astype(int) <= 2) &
        (rfm['F_score'].astype(int) <= 2)
    ],
    [
        'Champions',
        'Loyal Customers',
        'Potential Loyalists',
        'At Risk Customers',
        'Lost Customers'
    ],
    default='Regular Customers'
) 

rfm['segment'].value_counts()
```

```python
segment_analysis=rfm.groupby('segment').agg(
    no_of_customer=('transaction_id_x','nunique'),
    total_amount_per_segment=('amount_x','sum'),
    avg_amount_per_segment=('amount_x','mean')
)
segment_analysis['customer_contribution_per_segment']=(segment_analysis['no_of_customer']/no_of_unique_customer)*100
segment_analysis['amount_contribution_per_segment']=(segment_analysis['total_amount_per_segment']/segment_analysis['total_amount_per_segment'].sum())*100

segment_analysis
```
